# Helper Functions

```python
def prime_factorization(num):
    arr = []
    idx = 2

    while num > 1:
        if num % idx == 0:
            arr.append(idx)
            num = num // idx
            continue
        idx += 1

    return Counter(arr)

def load_alphabet():
    try:
        with open("alphabet.json", "r") as file:
            return json.load(file)
    except FileNotFoundError:
        print("Error: alphabet.json file not found ❌")
        return {}
    except json.JSONDecodeError:
        print("Error: alphabet.json is not valid JSON ❌")
        return {}
    
def get_inverse_brute_force(num, n=26):
    if gcd_euclid_fast(num, n) != 1:
        return -1
    
    for i in range(1, n):
        if (num * i) % n == 1:
            return i
        
    return -1

alph = load_alphabet()
alph_rev = {idx: char for char, idx in alph.items()}
```

# Greatest Common Divisor

```python
def gcd_prime_factorization(a, b):
    apf = prime_factorization(a)
    bpf = prime_factorization(b)
    gcd = 1

    for factor, count in apf.items():
        if factor in bpf:
            count_b = bpf.get(factor, 0)
            shared_counter = min(count, count_b)
            gcd *= (factor ** shared_counter)

    return gcd

def gcd_euclid(a, b):
    if b == 0:
        return a
    
    if a == 0:
        return b
    
    # Base case
    if a == b:
        return a
    
    if a > b:
        return gcd_euclid(a - b, b)
    
    return gcd_euclid(a, b - a)

def gcd_euclid_fast(a, b):
    if b == 0:
        return a
    return gcd_euclid_fast(b, a % b)
```

# Least Common Multiple

```python
def lcm_prime_factorization(a, b):
    apf = prime_factorization(a)
    bpf = prime_factorization(b)
    factors = set(apf.keys()) | set(bpf.keys())
    lcm = 1

    for factor in factors:
        count_a = apf.get(factor, 0)
        count_b = bpf.get(factor, 0)
        max_count = max(count_a, count_b)
        lcm *= (factor ** max_count)

    return lcm

def lcm_gcd(a, b):
    return abs(a) * (abs(b) // gcd_euclid_fast(a, b))
```

### Caesar's Encoding

```python
def caesar_encoding(plain_text, offset=3, decipher=False, mod=26):
    if not plain_text:
        return None
    
    shift = -offset if decipher else offset
    txt = []

    for char in plain_text:
        lower_char = char.lower()
        if lower_char in alph:
            i = alph[lower_char]
            j = (i + shift) % mod
            new_char = alph_rev[j]
            txt.append(new_char.upper() if char.isupper() else new_char)
        else:
            txt.append(char)

    cipher_text = "".join(txt)

    return cipher_text
        
```
